```python
import pandas as pd
```

```python
import glob
```

```python
import scanpy as sc
```

```python
h5ad_files = glob.glob("k_fold_mode/train_val/*/*.h5ad")
```

```python
all_cell_ids = []
```

```python
for h5ad_file in h5ad_files:
    ad = sc.read_h5ad(h5ad_file)
    cell_ids = list(ad.obs["soma_joinid"])
    all_cell_ids = all_cell_ids + cell_ids
```

```python
meta_df = pd.read_parquet("/home/gangcai/database/public_db/CZCELLxGENE/whole_datasets/CZCELLxGENE_Human_All/normal_final/meta_data/CZCELLxGENE_ScAgeClock_meta_filtered_and_splitted.parquet")
```

```python
meta_df_s = meta_df[meta_df["soma_joinid"].isin(all_cell_ids)]
```
